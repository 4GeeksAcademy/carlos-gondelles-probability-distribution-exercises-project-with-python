# Problemas de distribución de probabilidad

## Distribución binomial

En teoría de la probabilidad y estadística, la distribución binomial es la distribución de probabilidad discreta que da sólo dos resultados posibles en un experimento; por ejemplo, al lanzar una moneda, puede salir cara o cruz.

```python
from scipy.stats import binom

# Definir todas las funciones de probabilidad relacionadas con esta distribución.

def dbinom(x, size, prob = 0.5):
    """
    Calcula la estimación puntual de la distribución binomial.
    """
    result = binom.pmf(k = x, n = size, p = prob, loc = 0)

    return result

def pbinom(q, size, prob = 0.5):
    """
    Calcula el acumulado de la distribución binomial.
    """
    result = binom.cdf(k = q, n = size, p = prob, loc = 0)

    return result

def qbinom(p, size, prob = 0.5):
    """
    Calcula la función cuantil a partir de la distribución binomial.
    """
    result = binom.ppf(q = p, n = size, p = prob, loc = 0)

    return result

def rbinom(n, size, prob = 0.5):
    """
    Genera variables aleatorias a partir de la distribución binomial.
    """
    result = binom.rvs(n = size, p = prob, size = n)

    return result
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

np.random.seed(42)

# Parámetros de la distribución binomial
n = 10
p = 0.2

# Valores posibles de la variable aleatoria
x = np.arange(0, n + 1)

# Funciones principales
pmf = binom.pmf(x, n=n, p=p)
cdf = binom.cdf(x, n=n, p=p)

# Cálculos puntuales
probabilidad_2 = binom.pmf(2, n=n, p=p)
acumulada_2 = binom.cdf(2, n=n, p=p)
cuantil_09 = binom.ppf(0.9, n=n, p=p)

# Muestra aleatoria
muestra = binom.rvs(n=n, p=p, size=2000)

print(f"Probabilidad de que Binomial(10, 0.2) tome el valor 2: {probabilidad_2}")
print(f"Probabilidad acumulada hasta 2 en Binomial(10, 0.2): {acumulada_2}")
print(f"Valor cuyo acumulado alcanza 0.9 en Binomial(10, 0.2): {cuantil_09}")

fig, axis = plt.subplots(3, 1, figsize=(8, 10))

# PMF
axis[0].stem(x, pmf, basefmt=" ")
axis[0].set_title("Función de masa de probabilidad (PMF) - Binomial(10, 0.2)")
axis[0].set_xlabel("Número de éxitos")
axis[0].set_ylabel("Probabilidad")
axis[0].grid(alpha=0.3)

# CDF
axis[1].step(x, cdf, where="mid")
axis[1].scatter(x, cdf)
axis[1].set_title("Función de distribución acumulada (CDF) - Binomial(10, 0.2)")
axis[1].set_xlabel("Número de éxitos")
axis[1].set_ylabel("Probabilidad acumulada")
axis[1].grid(alpha=0.3)

# Simulación
axis[2].hist(muestra, bins=np.arange(-0.5, n + 1.5, 1), density=True, edgecolor="black")
axis[2].set_title("Simulación de 2000 valores - Binomial(10, 0.2)")
axis[2].set_xlabel("Número de éxitos")
axis[2].set_ylabel("Frecuencia relativa")
axis[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### Interpretación

- La **PMF** muestra la probabilidad de cada número posible de éxitos.
- La **CDF** muestra la probabilidad acumulada hasta cada valor.
- La simulación permite comprobar visualmente qué valores aparecen con más frecuencia en esta distribución.

## Ejercicio 1

En este apartado se trabajan dos distribuciones continuas: la normal y la uniforme.

Como son distribuciones continuas, en este caso no se utiliza una función de masa de probabilidad, sino:

- **PDF**: función de densidad de probabilidad;
- **CDF**: función de distribución acumulada.

El objetivo es calcular estas funciones y representarlas gráficamente para observar cómo cambia su forma según el tipo de distribución.

```python
from scipy.stats import norm

# Distribucion normal

def dnorm(x, mean = 0, sd = 1):

    result = norm.pdf(x, loc = mean, scale= sd)

    return result


def pnorm(x, mean = 0, sd = 1):
   
    result = norm.cdf(x, loc = mean, scale = sd)

    return result

def qnorm(p, mean = 0, sd = 1):

    result = norm.ppf(q = p, loc = mean, scale = sd)

    return result

def rvsnorm(n, mean = 0, sd = 1):
   
    result = norm.rvs(loc = mean, scale = sd, size= n)

    return result


# Distribuciones Uniformes

from scipy.stats import uniform

def duniform(x, mean = 0, sd = 1):

    result = uniform.pdf(x, loc = mean, scale= sd)

    return result


def puniform(x, mean = 0, sd = 1):
   
    result = uniform.cdf(x, loc = mean, scale = sd)

    return result

def quniform(p, mean = 0, sd = 1):

    result = uniform.ppf(q = p, loc = mean, scale = sd)

    return result

def rvsuniform(n, mean = 0, sd = 1):

    result = uniform.rvs(loc = mean, scale = sd, size= n)

    return result
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform

# Parámetros para la normal
media = 0
desviacion = 1

# Parámetros para la uniforme
inicio = 0
ancho = 10

# Rango de valores
x_normal = np.linspace(-4, 4, 500)
x_uniforme = np.linspace(-2, 12, 500)

# Funciones de la normal
pdf_normal = norm.pdf(x_normal, loc=media, scale=desviacion)
cdf_normal = norm.cdf(x_normal, loc=media, scale=desviacion)

# Funciones de la uniforme
pdf_uniforme = uniform.pdf(x_uniforme, loc=inicio, scale=ancho)
cdf_uniforme = uniform.cdf(x_uniforme, loc=inicio, scale=ancho)

# Cálculos puntuales de ejemplo
densidad_normal_0 = norm.pdf(0, loc=media, scale=desviacion)
acumulada_normal_0 = norm.cdf(0, loc=media, scale=desviacion)

densidad_uniforme_5 = uniform.pdf(5, loc=inicio, scale=ancho)
acumulada_uniforme_5 = uniform.cdf(5, loc=inicio, scale=ancho)

print(f"Densidad normal en x = 0: {densidad_normal_0}")
print(f"Probabilidad acumulada normal en x = 0: {acumulada_normal_0}")
print(f"Densidad uniforme en x = 5: {densidad_uniforme_5}")
print(f"Probabilidad acumulada uniforme en x = 5: {acumulada_uniforme_5}")

fig, axis = plt.subplots(2, 2, figsize=(10, 8))

# Normal PDF
axis[0, 0].plot(x_normal, pdf_normal)
axis[0, 0].set_title("PDF - Distribución normal")
axis[0, 0].set_xlabel("x")
axis[0, 0].set_ylabel("Densidad")
axis[0, 0].grid(alpha=0.3)

# Normal CDF
axis[0, 1].plot(x_normal, cdf_normal)
axis[0, 1].set_title("CDF - Distribución normal")
axis[0, 1].set_xlabel("x")
axis[0, 1].set_ylabel("Probabilidad acumulada")
axis[0, 1].grid(alpha=0.3)

# Uniforme PDF
axis[1, 0].plot(x_uniforme, pdf_uniforme)
axis[1, 0].set_title("PDF - Distribución uniforme")
axis[1, 0].set_xlabel("x")
axis[1, 0].set_ylabel("Densidad")
axis[1, 0].grid(alpha=0.3)

# Uniforme CDF
axis[1, 1].plot(x_uniforme, cdf_uniforme)
axis[1, 1].set_title("CDF - Distribución uniforme")
axis[1, 1].set_xlabel("x")
axis[1, 1].set_ylabel("Probabilidad acumulada")
axis[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
